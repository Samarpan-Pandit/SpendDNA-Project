```python
def extract_vendor(description):

    if pd.isna(description):
        return "Unknown"

    desc = str(description).upper()

    # -------------------------------
    # Cash Withdrawal
    # -------------------------------
    if "ATM-WDL" in desc:
        return "Cash Withdrawal"

    # -------------------------------
    # Personal Transfers
    # -------------------------------
    if ("UPI-AMAN" in desc or
        "UPI-PRIYA" in desc or
        "UPI-ANKIT" in desc or
        "UPI-VIKAS" in desc):
        return "P2P Transfer"

    # -------------------------------
    # Salary
    # -------------------------------
    if "SALARY" in desc:
        return "Salary"

    # -------------------------------
    # Rent
    # -------------------------------
    if "RENT" in desc:
        return "Rent"

    # -------------------------------
    # Instamart (CHECK FIRST)
    # -------------------------------
    if "INSTAMART" in desc:
        return "Instamart"

    # -------------------------------
    # Vendor Dictionary Matching
    # -------------------------------
    for vendor, keywords in vendor_dictionary.items():

        for keyword in keywords:

            if keyword in desc:
                return vendor

    return "Uncategorised"
```

```python
df["vendor_clean"] = df["Description"].apply(extract_vendor)
```

```python
# AI-assisted:
# Total Credits

total_credit = credit_df["Amount"].sum()

# Total Debits

total_debit = debit_df["Amount"].sum()

# Net Savings

net_savings = total_credit - total_debit

# Savings Rate

if total_credit != 0:
    savings_rate = (net_savings / total_credit) * 100
else:
    savings_rate = 0

# Transaction Count

total_transactions = len(df)

# Unique Vendors

unique_vendors = df["vendor_clean"].nunique()

# Unique Categories

unique_categories = df["category"].nunique()
```

#Feature 5 – Monthly Trend Analysis

```python
# ==========================================================
# MONTHLY CATEGORY SPENDING PIVOT
# ==========================================================

monthly_pivot = pd.pivot_table(
    consumer_spending,
    values="Amount",
    index="category",
    columns="month_name",
    aggfunc="sum",
    fill_value=0
)

# Arrange months in chronological order
monthly_pivot = monthly_pivot.reindex(columns=month_order)

print("="*90)
print("CATEGORY × MONTH SPENDING")
print("="*90)

print(monthly_pivot.round(2))
```

```python
print("="*90)
print("MONTHLY SPENDING TABLE")
print("="*90)

for category in monthly_pivot.index:

    print(f"\n{category}")

    for month in month_order:

        amount = monthly_pivot.loc[category, month]

        print(f"{month:<10} ₹ {amount:>10,.2f}")
```

```python
monthly_total = (
    consumer_spending
    .groupby("month_name")["Amount"]
    .sum()
    .reindex(month_order)
)

print("="*90)
print("TOTAL SPENDING PER MONTH")
print("="*90)

print(monthly_total)
```

```python
print("="*90)
print("MONTHLY SPENDING TREND")
print("="*90)

maximum = monthly_total.max()

for month, amount in monthly_total.items():

    bar = ascii_bar(amount, maximum)

    print(f"{month:<10} {bar:<30} ₹ {amount:,.2f}")
```

```python
growth = {}

for category in monthly_pivot.index:

    january = monthly_pivot.loc[category, "January"]
    june = monthly_pivot.loc[category, "June"]

    if january == 0:
        growth[category] = np.nan
    else:
        growth[category] = ((june - january) / january) * 100
```

```python
growth_df = pd.DataFrame({

    "Category": growth.keys(),
    "Growth (%)": growth.values()

})

growth_df = growth_df.dropna()

growth_df = growth_df.sort_values(
    by="Growth (%)",
    ascending=False
)

print("="*90)
print("CATEGORY GROWTH")
print("="*90)

print(growth_df.round(2))
```

```python
highest_growth = growth_df.iloc[0]

biggest_decline = growth_df.iloc[-1]

print("="*90)
print("TREND ANALYSIS")
print("="*90)

print(f"Highest Growth Category : {highest_growth['Category']}")
print(f"Growth Percentage       : {highest_growth['Growth (%)']:.2f}%")

print()

print(f"Biggest Decline Category: {biggest_decline['Category']}")
print(f"Decline Percentage      : {biggest_decline['Growth (%)']:.2f}%")
```

#Feature 6 – Time of Day Analysis

```python
#Create Hour × Category Pivot Table

hourly_pivot = pd.pivot_table(
    consumer_spending,
    values="Amount",
    index="category",
    columns="hour",
    aggfunc="sum",
    fill_value=0
)

# Ensure all 24 hours exist
hourly_pivot = hourly_pivot.reindex(columns=range(24), fill_value=0)

print("="*100)
print("CATEGORY × HOUR SPENDING MATRIX")
print("="*100)

print(hourly_pivot.round(2))
```

```python
#Total Spending Per Hour
hourly_total = (
    consumer_spending
    .groupby("hour")["Amount"]
    .sum()
    .reindex(range(24), fill_value=0)
)
print("="*100)
print("TOTAL SPENDING BY HOUR")
print("="*100)
print(hourly_total)
```

```python
#ASCII Hourly Trend
print("="*100)
print("HOURLY SPENDING TREND")
print("="*100)

maximum = hourly_total.max()

for hour, amount in hourly_total.items():

    bar = ascii_bar(amount, maximum)

    print(f"{hour:02d}:00  {bar:<30} ₹ {amount:,.2f}")
```

```python
#Peak Hour Of Every Category
print("="*100)
print("PEAK HOUR OF EVERY CATEGORY")
print("="*100)

peak_hours = {}

for category in hourly_pivot.index:

    peak_hour = hourly_pivot.loc[category].idxmax()
    amount = hourly_pivot.loc[category].max()

    peak_hours[category] = peak_hour

    print(f"{category:<20} {peak_hour:02d}:00   ₹ {amount:,.2f}")
```

```python
#Food Delivery Late Night Analysis
food = consumer_spending[
    consumer_spending["category"] == "Food Delivery"
].copy()

total_food_orders = len(food)

late_night_orders = food[
    (food["hour"] >= 21) |
    (food["hour"] <= 1)
]

late_orders = len(late_night_orders)

if total_food_orders > 0:
    late_percentage = (late_orders / total_food_orders) * 100
else:
    late_percentage = 0

print("="*100)
print("FOOD DELIVERY ANALYSIS")
print("="*100)

print(f"Total Food Orders       : {total_food_orders}")
print(f"Late Night Orders       : {late_orders}")
print(f"Late Night Percentage   : {late_percentage:.2f}%")
```

```python
# AI-assisted:
#Cafe Morning Analysis
cafe = consumer_spending[
    consumer_spending["category"] == "Cafe"
]

total_cafe = len(cafe)

morning_cafe = cafe[
    (cafe["hour"] >= 8) &
    (cafe["hour"] <= 11)
]

morning_orders = len(morning_cafe)

if total_cafe > 0:
    morning_percentage = (morning_orders / total_cafe) * 100
else:
    morning_percentage = 0

print("="*100)
print("CAFE ANALYSIS")
print("="*100)

print(f"Total Cafe Visits     : {total_cafe}")
print(f"Morning Visits        : {morning_orders}")
print(f"Morning Percentage    : {morning_percentage:.2f}%")
```

```python
# AI-assisted:
#Spending By Time Period
def get_time_period(hour):

    if 5 <= hour <= 11:
        return "Morning"

    elif 12 <= hour <= 16:
        return "Afternoon"

    elif 17 <= hour <= 20:
        return "Evening"

    else:
        return "Night"


consumer_spending["time_period"] = consumer_spending["hour"].apply(get_time_period)

time_period_summary = (
    consumer_spending
    .groupby("time_period")["Amount"]
    .sum()
)

print("="*100)
print("SPENDING BY TIME PERIOD")
print("="*100)

print(time_period_summary)
```

```python
print("="*100)
print("TIME PERIOD ANALYSIS")
print("="*100)

maximum = time_period_summary.max()

for period, amount in time_period_summary.items():

    bar = ascii_bar(amount, maximum)

    print(f"{period:<12} {bar:<30} ₹ {amount:,.2f}")
```

#Feature 7 : Anomaly Detection

```python
anomaly_df = consumer_spending.copy()
```

```python
# Mean Amount of each Category

anomaly_df["category_mean"] = (
    anomaly_df
    .groupby("category")["Amount"]
    .transform("mean")
)
```

```python
# Standard Deviation of each Category

anomaly_df["category_std"] = (
    anomaly_df
    .groupby("category")["Amount"]
    .transform("std")
)
```

```python
anomaly_df["category_std"] = anomaly_df["category_std"].replace(0, np.nan)
```

```python
# z-score

anomaly_df["z_score"] = (

    anomaly_df["Amount"] -
    anomaly_df["category_mean"]

) / anomaly_df["category_std"]
```

```python
anomaly_df["z_score"] = anomaly_df["z_score"].fillna(0)
```

```python
# Transactions having z-score > 2

anomalies = anomaly_df[
    anomaly_df["z_score"] > 2
].copy()
```

```python
anomalies = anomalies.sort_values(
    by="z_score",
    ascending=False
)
```

```python
anomalies.reset_index(
    drop=True,
    inplace=True
)
```
